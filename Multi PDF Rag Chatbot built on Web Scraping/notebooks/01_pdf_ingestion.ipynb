```python
from langchain_huggingface import HuggingFaceEmbeddings
```

3. Create the embedding model

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
text = "The phone has a 5000 mAh battery."

vector = embeddings.embed_query(text)

print(type(vector))
print("Vector dimension:", len(vector))
print(vector[:10])
```

```python
chunk_texts = [chunk.page_content for chunk in chunks]

chunk_vectors = embeddings.embed_documents(chunk_texts)

print("Number of chunks:", len(chunks))
print("Number of vectors:", len(chunk_vectors))
print("Vector dimension:", len(chunk_vectors[0]))
```

## Step 4 — FAISS Vector Index

```python
from langchain_community.vectorstores import FAISS
```

Create the FAISS index

```python
pip install faiss-cpu
```

```python
vector_db = FAISS.from_documents(
    chunks,
    embeddings
)
```

Check the FAISS index

```python
print(vector_db.index.ntotal)
```

```python
len(chunks)
```

```python
query = "Where is Econocom situated?"
```

```python
results = vector_db.similarity_search(
    query,
    k=3
)
```

```python
for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content[:500])
    print("Metadata:", doc.metadata)
```

We can also retrieve scores:

```python
results_with_scores = vector_db.similarity_search_with_score(
    query,
    k=3
)
```

```python
for doc, score in results_with_scores:
    print("\nScore:", score)
    print(doc.page_content[:300])
```

```python
vector_db.save_local("../faiss")
```

## LangChain Ollama integration

```python
from langchain_ollama import ChatOllama
```

Create the LLM:

```python
llm = ChatOllama(
    model="tinyllama",
    temperature=0
)
```

Before connecting it to RAG, test

```python
response = llm.invoke(
    "What is machine learning? Answer in two sentences."
)

print(response.content)
```

Now connect the Retriever + LLM

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_template("""
You are an e-commerce product assistant.

Answer the question using ONLY the context provided below.

If the answer cannot be found in the context, say:
"I couldn't find that information in the provided documents."

Context:
{context}

Question:
{question}

Answer:
""")
```

```python
## Retrieve the documents
```

```python
question = "What is the registered office of Econocom?"
```

```python
docs = retriever.invoke(question)
```

```python
context = "\n\n".join(
    doc.page_content for doc in docs
)
```

## Build the prompt

```python
formatted_prompt = prompt.invoke({
    "context": context,
    "question": question
})
```

```python
print(formatted_prompt)
```

Send that to TinyLlama

```python
response = llm.invoke(formatted_prompt)

print(response.content)
```

## Let's make it into one function

```python
def ask_rag(question):
    docs = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content for doc in docs
    )

    formatted_prompt = prompt.invoke({
        "context": context,
        "question": question
    })

    response = llm.invoke(formatted_prompt)

    return response.content, docs
```

```python
answer, sources = ask_rag(
    "What is the registered office of Econocom?"
)

print(answer)
```

And inspect sources:

```python
for doc in sources:
    print(doc.metadata)
```

```python
answer, sources = ask_rag(
    "What is the registered office of Econocom?"
)

print("ANSWER:")
print(answer)

print("\nSOURCES:")

for doc in sources:
    print(
        doc.metadata.get("source"),
        "Page:",
        doc.metadata.get("page")
    )
```

## Ask a question that isn't in your PDFs.

```python
answer, sources = ask_rag(
    "Who is the CEO of Tesla?"
)

print(answer)
```
